```python
month=int(input("Enter the Month 1 to 12 "))
match month:
  case 1: print("January")
  case 2: print("February")
  case 3: print("March")
  case 4: print("April")
  case 5: print("May")
  case 6: print("June")
  case 7: print("Jully")
  case 8: print("August")
  case 9: print("September")
  case 10: print("October")
  case 11: print("November")
  case 12: print("December")
  case _: print("Invalid Month (Please Enter the Month Valid)")

```

```python
num1=int(input("Enter the Number 1 "))
num2=int(input("enter the number 2 "))
answer=input("enter the value +,-,*,/,% ")
match answer:
  case "+": print(num1+num2)
  case "-": print(num1-num2)
  case "*": print(num1*num2)
  case "/": print(num1/num2)
  case "%": print(num1%num2)
  case _: print(" Invalid symbles (enter the value +,-,*,/,%)")

```
