```python
# Functions - A Functions is a block of code which only runs when its called. you can pass data, known as parameters, into a functions.
# A fucntions can return data as a result.
```

```python
def name():
  print("Helo chennai")
name()

```

```python
def total():
  a=8
  b=12
  c=a+b
  print(c)
total()
```

```python
# with parameters without return type
def largevalue(a,b,c,d,e):
  sum=a+b+c+d+e
  average=sum/5
  print(sum)
  print(average)
largevalue(89,98,90,97,76)
```

```python
# without parameter with written type
def add():
  a=8
  b=8
  return a+b
print(add())
```

```python
# with parameter with return
def total(a,b,c,d,e):
  sum=a+b+c+d+e
  return sum
print(total(89,98,90,97,76))
```

```python
def weekday(*day):
 print(day)
 print(day[0])
 print(day[1])
 print(day[2])
 print(day[3])
 print(day[4])
weekday("monday", "tuesday", "wednessday", "thursday", "friday")
```

```python
def day(c1,c2,c3,c4,c5):
  print(c1)
  print(c2)
  print(c3)
  print(c4)
  print(c5)
day("monday", "tuesday", "wednessday", "thursday", "friday")
```

```python
#** arguments
def day(**day):
  print(day)
  print(day["c1"])
  print(day["c2"])
  print(day["c3"])
  print(day["c4"])
  print(day["c5"])
day(c1="monday", c2="tuesday", c3="wednessday", c4="thursday", c5="friday")
```
