```python
# Dot product
#Big Data Example (User–Movie Recommendation System)
#✔ 5 Users
#✔ 6 Movies
#✔ 3 Features (Action, Comedy, Drama)
```

```python
import numpy as np
```

```python
#User preference matrics(5 users, 3 preferance)
A=np.array([

[4,1,2],
[2,5,1],
[1,1,5],
[3,4,2],
[5,0,1],
])
#Movie preference matris(6 movies, 3 feautures)
B=np.array([
    [5,0,3],
    [1,5,0],
    [2,1,4],
    [4,3,1],
    [0,4,5],
    [3,2,2],
])
Scores=np.dot(A,B.T)
print(Scores)
```

```python
#Music Streaming Platform – Sample Practice Data
#User Preference Matrix (8 users × 4 features)
Users=np.array([
    [5, 2, 1, 3],   # User 1
    [1, 4, 3, 2],   # User 2
    [3, 5, 0, 1],   # User 3
    [2, 1, 4, 5],   # User 4
    [4, 0, 2, 3],   # User 5
    [0, 3, 5, 1],   # User 6
    [2, 4, 1, 0],   # User 7
    [5, 1, 0, 4]    # User 8
])
#Song Feature Matrix (10 songs × 4 features)
songs = np.array([
    [4, 3, 0, 2],   # Song 1
    [1, 5, 2, 0],   # Song 2
    [3, 0, 4, 1],   # Song 3
    [5, 2, 1, 3],   # Song 4
    [0, 4, 5, 1],   # Song 5
    [2, 1, 3, 4],   # Song 6
    [4, 0, 2, 5],   # Song 7
    [1, 3, 0, 4],   # Song 8
    [3, 2, 5, 0],   # Song 9
    [0, 1, 4, 3]    # Song 10
])
Scores=np.dot(Users,songs.T)
print(Scores)
```

```python
# CUMULATIVE PRODUCT.NP.CUMPROD()
#quantity = [5, 3, 2, 4, 6, 2, 3, 5, 4, 2,
           # 7, 3, 2, 5, 4, 3, 6, 2, 3, 4]
A=np.array( [5, 3, 2, 4, 6, 2, 3, 5, 4, 2,
            7, 3, 2, 5, 4, 3, 6, 2, 3, 4]
)
print(np.cumprod(A))
```
