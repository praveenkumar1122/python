```python
# joint set union and update
# intersection(), difference(), symmetric_difference()
S1={"Python", "SQL", "Excel", "Power BI"}
S2={8000, 10000, 6000, 30000}
answer=S1.union(S2)
print(answer)
```

```python
S1={"Python", "SQL", "Excel", "Power BI"}
S2={8000, 10000, 6000, 30000}
answer= S1 | S2
print(answer)
```

```python
S1={"Python", "SQL", "Excel", "Power BI"}
S2={8000, 10000, 6000, 30000}
S3={"BBA", "MBA", "BSC"}
S4={11000, 21000}
answer=S1.union(S2, S3, S4)
print(answer)
```

```python
S1={"Python", "SQL", "Excel", "Power BI"}
S2={8000, 10000, 6000, 30000}
S3={"BBA", "MBA", "BSC"}
S4={11000, 21000}
answer=S1 | S2 | S3 | S4
print(answer)
```
