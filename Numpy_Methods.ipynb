```python
arr=np.array([[1,3,6],[5,7,9],[0,6,8]])
print(arr)
print(arr.shape)
print(arr.ndim)
```

```python
#aray indexing and slicing
arr=np.array([15,20,30,40,55,80])
print(arr[0])
print(arr[-1])
print(arr[1:4])
print(arr[-2])
```

```python
# array operations
#element wise operations
a=np.array([1,2,4,8])
b=np.array([8,3,7,9])

print(a+b)
print(a*b)
print(a**b)
```

```python
# mathamatical functions
arr=np.array([12,12,13,14])
print(np.sum(arr))
print(np.average(arr))
print(np.mean(arr))
print(np.max(arr))
print(np.min(arr))
print(np.sqrt(arr))
```

```python
#Reshaping arreas
arr=np.arange(6)
print(arr)
newarr=arr.reshape(2,3)
print(newarr)
```

```python
# comparison and logical operator
arr=np.array([10,20,30,40,50])
print(arr>=30)
print(arr<=30)

print(arr[arr>=30])
print(arr[arr<30])
```

```python
# copy vs view
a=np.array([1,2,3,4,5])
b=a.copy()

print(a)
print(b)

print(b[0])
b[0]=10
print(b[0])

print(b)

```
